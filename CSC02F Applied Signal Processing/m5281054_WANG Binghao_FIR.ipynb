```python
x = [0, 1, 1, 1, 0, 0, 0, 0]
h = [-1, 3]
```

```python
y = []
for n in range(len(x) - len(h) + 1):
    # Compute the dot product of the overlapping segments
    result = sum(x[n + k] * h[k] for k in range(len(h)))
    y.append(result)
```

```python
print("FIR Result:", y)
```
