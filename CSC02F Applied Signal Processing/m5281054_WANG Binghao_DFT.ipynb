```python
import math
```

```python
import cmath
```

```python
def compute_dft(x):
    N = len(x)
    X = []
    for k in range(N):
        sum_value = 0
        for n in range(N):
            angle = -2 * math.pi * k * n / N
            sum_value += x[n] * cmath.exp(complex(0, angle))
        X.append(sum_value)
    return X
```

```python
x = [complex(0, 1), 3, 0, complex(1, 1)]
```

```python
X = compute_dft(x)
```

```python
print("DFT Result:", X)
```
