```python
import cmath
```

```python
def fft(x):
    N = len(x)
    if N <= 1:
        return x
    even = fft(x[0::2])  # 偶数项
    odd = fft(x[1::2])   # 奇数项
    T = [cmath.exp(-2j * cmath.pi * k / N) * odd[k] for k in range(N // 2)]
    return [even[k] + T[k] for k in range(N // 2)] + \
           [even[k] - T[k] for k in range(N // 2)]
```

```python
x = [1, 0, 2j, 4]
```

```python
X = fft(x)
```

```python
print("FFT Result:", X)
```
